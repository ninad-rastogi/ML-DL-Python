**Logistic Regression Convergence Warning**

While fitting the Logistic Regression model using:

```python
clf_lrs_multi = LogisticRegression()
clf_lrs_multi.fit(X_multi, y)
```

the following warning was generated:

```text
ConvergenceWarning: lbfgs failed to converge after 100 iteration(s)
STOP: TOTAL NO. OF ITERATIONS REACHED LIMIT
```

---

1. **Why This Warning Occurs**

Logistic Regression uses an optimization algorithm to find the best coefficients for the model.

By default, Scikit-Learn uses the `lbfgs` solver, which iteratively updates the model coefficients until convergence is achieved.

Convergence means:
- the model parameters have stabilized
- the loss function has been minimized
- the optimal solution has been reached

However, the optimization process stopped before convergence because the default maximum number of iterations (`max_iter=100`) was reached.

---

2. **Mathematical Theory Behind Logistic Regression**

Logistic Regression predicts probabilities using the sigmoid function:

$$
P(Y=1)=\frac{1}{1+e^{-z}}
$$

Where:

$$
z=\beta_0+\beta_1x_1+\beta_2x_2+...+\beta_nx_n
$$

Here:
- $\beta_0$ = intercept
- $\beta_1, \beta_2, ...$ = feature coefficients
- $x_1, x_2, ...$ = feature values

The optimizer continuously updates the coefficients to minimize the loss function.

If convergence is not achieved within the allowed iterations, the warning appears.

---

3. **Common Reasons for This Warning**

| Cause                   | Explanation                                   |
|-------------------------|-----------------------------------------------|
| Features are not scaled | Large feature magnitudes slow optimization    |
| Complex dataset         | More iterations are needed                    |
| Multicollinearity       | Highly correlated features create instability |
| Too many features       | Optimization becomes harder                   |
| Default iteration limit | `max_iter=100` may not be sufficient          |

---

4. **Why Feature Scaling Helps**

Different features may have very different numerical ranges.

Example:

| Feature | Range               |
|---------|---------------------|
| Age     | 20 to 60            |
| Salary  | 10,000 to 1,000,000 |

Large differences in scale make gradient-based optimization difficult.

`StandardScaler()` standardizes features using:

$$
z = \frac{x - \mu}{\sigma}
$$

Where:
- $\mu$ = mean
- $\sigma$ = standard deviation

After scaling:
- mean becomes 0
- standard deviation becomes 1

This helps the optimizer converge faster and more efficiently.

---

5. **Solution**

The warning can usually be resolved by:

1. Scaling the features
2. Increasing the maximum number of iterations

---

6. **Code Implementation**

**Step 1:** Import Required Libraries


## 4.1. Setting-up the boundary conditions for a house being **Sold** or **Not Sold**

* By default, the boundary condition is `0.5` which means if the probability is < 0.5, the house can not be sold within 3 months.
* Cost associated with `False Positive` & `False Negative` is different, therefore we may want to choose a different boundary condition.
* **Example:**

**Confusion Matrix Analogy Using Facial Recognition Security Lock**

Suppose there is a **facial recognition security lock** installed at the entrance of a highly secure military research facility.

The system must decide:

- **Allow Entry** → Person is authenticated
- **Deny Entry** → Person is rejected

The actual reality can be:

- Person truly has access
- Person truly does not have access

This creates four possible outcomes in the confusion matrix.

---

**Understanding Positive and Negative Classes**

In this scenario:

| Term      | Meaning                |
|-----------|------------------------|
| Positive  | Person is authorized   |
| Negative  | Person is unauthorized |

The AI model predicts:
- Authorized
- Unauthorized

---

1. **True Positive (TP)**

**Situation**

An authorized person comes to the door, and the system correctly grants access.

**Example**

- Employee has valid access
- Facial recognition identifies correctly
- Door opens

**Result**

✅ Correct decision

✅ Secure and convenient

---

2. **True Negative (TN)**

**Situation**

An unauthorized person tries to enter, and the system correctly blocks access.

**Example**

- Intruder attempts entry
- System identifies them as unauthorized
- Door remains locked

**Result**

✅ Correct decision

✅ Security maintained

---

3. **False Positive (FP)**

**Situation**

An unauthorized person is mistakenly identified as authorized.

**Example**

- Intruder attempts entry
- Facial recognition incorrectly authenticates them
- Door opens

**Result**

❌ Extremely dangerous

❌ Security breach occurs

This is the most critical error in this scenario.

---

4. **False Negative (FN)**

**Situation**

An authorized person is mistakenly rejected.

**Example**

- Valid employee attempts entry
- Facial recognition fails
- Door remains locked

**Result**

❌ Inconvenient

❌ But less dangerous

The authorized person can:
- try again
- use another authentication method
- contact security

---

**Why False Positives Are More Dangerous Here**

In this system:

| Error Type     | Consequence                             |
|----------------|-----------------------------------------|
| False Positive | Unauthorized person gains access        |
| False Negative | Authorized person is temporarily denied |

A False Negative causes inconvenience.

A False Positive causes a security breach.

Therefore, the system should prioritize minimizing:

$$
False\ Positives
$$

even if that slightly increases False Negatives.

---

**Boundary Condition Analogy**

The facial recognition model internally produces a probability score.

**Example:**

| Person                   | Probability of Being Authorized |
|--------------------------|---------------------------------|
| Employee A               | 0.95                            |
| Employee B               | 0.82                            |
| Unknown Person           | 0.12                            |
| Similar-looking Intruder | 0.61                            |

The system sets a decision threshold (boundary condition).

**Example:**

$$
Threshold = 0.90
$$

Decision rule:

- Probability ≥ 0.90 → Allow access
- Probability < 0.90 → Deny access

---

**Effect of a Strict Boundary**

If the threshold is very high:

| Advantage               | Disadvantage                |
|-------------------------|-----------------------------|
| Fewer intruders allowed | More genuine users rejected |

This reduces:
- False Positives

But increases:
- False Negatives

---

**Effect of a Loose Boundary**

If the threshold is low:

| Advantage                   | Disadvantage             |
|-----------------------------|--------------------------|
| More genuine users accepted | More intruders may enter |

This reduces:
- False Negatives

But increases:
- False Positives

---

**Real-World Design Decision**

**For a high-security military facility:**

The system is intentionally designed with a strict boundary because:

$$
False\ Positive\ Risk \gg False\ Negative\ Risk
$$

**Meaning:**

Allowing an intruder inside is far more dangerous than temporarily rejecting a legitimate employee.

---

**Complete Mapping to Confusion Matrix**

| Actual Condition    | Predicted Authorized  | Predicted Unauthorized |
|---------------------|-----------------------|------------------------|
| Authorized Person   | True Positive (TP)    | False Negative (FN)    |
| Unauthorized Person | False Positive (FP)   | True Negative (TN)     |

---

**Key Insight**

Confusion matrix is not just about accuracy.

It helps determine:
- which mistakes are more dangerous
- how to adjust decision boundaries
- which metric matters most in a real-world application

In security systems:
- minimizing False Positives is usually the top priority.

In medical diagnosis:
- minimizing False Negatives is often more important.

## 4.3. **Performance Measures in Confusion Matrix**

A Confusion Matrix is used to evaluate the performance of a classification model.

It compares:
- Actual values
- Predicted values

and summarizes the results into four categories.

---

### 4.3.1. **Confusion Matrix Structure**

| Actual / Predicted | Positive Prediction | Negative Prediction |
|--------------------|---------------------|---------------------|
| Actual Positive    | True Positive (TP)  | False Negative (FN) |
| Actual Negative    | False Positive (FP) | True Negative (TN)  |

---

### 4.3.2. **Understanding the Terms**

---


**True Positive (TP)**

The model correctly predicts the positive class.

**Example**
- Authorized person correctly allowed access
- Disease correctly detected

---

**True Negative (TN)**

The model correctly predicts the negative class.

**Example**
- Unauthorized person correctly rejected
- Healthy person correctly identified

---

**False Positive (FP)**

The model incorrectly predicts positive when the actual class is negative.

**Example**
- Intruder incorrectly granted access
- Healthy person incorrectly diagnosed with disease

Also called:

$$
Type\ I\ Error
$$

---

**False Negative (FN)**

The model incorrectly predicts negative when the actual class is positive.

**Example**
- Authorized person denied access
- Diseased patient incorrectly marked healthy

Also called:

$$
Type\ II\ Error
$$

---

## 4.3.3. **Why Performance Metrics Are Needed**

Accuracy alone is not always sufficient.

Example:
- In fraud detection, fraud cases may be very rare.
- A model predicting "No Fraud" for every transaction may still achieve high accuracy.

Therefore, additional performance measures are required.

---

## 4.3.4. **Accuracy**


Accuracy measures the overall correctness of the model.

Formula:

$$
Accuracy = \frac{TP + TN}{TP + TN + FP + FN}
$$

---

**Interpretation**

- Higher accuracy indicates better overall prediction.
- Useful when classes are balanced.

---

## 4.3.5. **Precision**

Precision measures how many predicted positive cases are actually positive.

Formula:

$$
Precision = \frac{TP}{TP + FP}
$$

---

**Interpretation**

High precision means:
- fewer False Positives
- positive predictions are reliable

---

**Importance**

Precision is important when False Positives are dangerous.

**Example Applications**

- Facial recognition security systems
- Spam filtering
- Fraud detection

---

## 4.3.6. **Recall (Sensitivity or True Positive Rate)**

Recall measures how many actual positive cases are correctly identified.

Formula:

$$
Recall = \frac{TP}{TP + FN}
$$

---

**Interpretation**

High recall means:
- fewer False Negatives
- most positive cases are detected

---

**Importance**

Recall is important when False Negatives are dangerous.

**Example Applications**

- Cancer detection
- Fraud detection
- Security threat detection

Missing a real positive case can be very harmful.

---

## 4.3.7. **Specificity (True Negative Rate)**

Specificity measures how well the model identifies negative cases.

Formula:

$$
Specificity = \frac{TN}{TN + FP}
$$

---

**Interpretation**

High specificity means:
- fewer False Positives
- negatives are correctly identified

---

## 4.3.8. **F1 Score**

F1 Score is the harmonic mean of Precision and Recall.

Formula:

$$
F1 = 2 \times \frac{Precision \times Recall}{Precision + Recall}
$$

---

**Interpretation**

F1 Score balances:
- Precision
- Recall

Useful when:
- class imbalance exists
- both FP and FN are important

---

## 4.3.9. **False Positive Rate (FPR)**

Measures how many negative cases are incorrectly predicted as positive.

Formula:

$$
FPR = \frac{FP}{FP + TN}
$$

---

**Importance**

Used in:
- ROC Curve analysis
- Threshold tuning

Lower FPR is usually preferred.

---

## 4.3.10. **False Negative Rate (FNR)**

Measures how many positive cases are incorrectly predicted as negative.

Formula:

$$
FNR = \frac{FN}{FN + TP}
$$

---

**Importance**

Critical in:
- medical diagnosis
- fraud detection
- threat detection

Lower FNR is usually preferred.

---

## 4.3.11. **Trade-off Between Precision and Recall**

Increasing one metric may reduce the other.

---

**Strict Classification Boundary**

- Fewer False Positives
- Higher Precision
- Lower Recall

---

**Loose Classification Boundary**

- Fewer False Negatives
- Higher Recall
- Lower Precision

---

## 4.3.12. **Which Metric Should Be Prioritized?**

Depends on the application.

| Application          | Important Metric  | Reason                                 |
|----------------------|-------------------|----------------------------------------|
| Medical Diagnosis    | Recall            | Missing disease is dangerous           |
| Spam Detection       | Precision         | Important emails should not go to spam |
| Security Lock System | Precision         | Unauthorized access is dangerous       |
| Fraud Detection      | Recall            | Fraud should not be missed             |

---

## 4.3.13. **Summary of Metrics**

| Metric      | Formula                             | Focus                               |
|-------------|-------------------------------------|-------------------------------------|
| Accuracy    | $(TP+TN)/(Total)$                   | Overall correctness                 |
| Precision   | $TP/(TP+FP)$                        | Reliability of positive predictions |
| Recall      | $TP/(TP+FN)$                        | Ability to detect positives         |
| Specificity | $TN/(TN+FP)$                        | Ability to detect negatives         |
| F1 Score    | Harmonic mean of Precision & Recall | Balance between Precision & Recall  |
| FPR         | $FP/(FP+TN)$                        | False alarm rate                    |
| FNR         | $FN/(FN+TP)$                        | Miss rate                           |

# 5. Linear Discriminant Analysis (LDA)

Linear Discriminant Analysis (LDA) is a supervised machine learning algorithm used for:
- classification
- dimensionality reduction

LDA tries to find a linear boundary that best separates different classes.

The main objective of LDA is:
- maximize separation between classes
- minimize variation within the same class

Unlike Logistic Regression, which directly predicts probabilities, LDA models the distribution of each class and then classifies data points.

---

**Working Principle**

LDA projects data onto a lower-dimensional space such that class separability becomes maximum.

It:
1. Computes class means
2. Measures spread within classes
3. Measures spread between classes
4. Finds the best linear combination of features for separation

---

**Applications of LDA**

- Face recognition
- Medical diagnosis
- Customer classification
- Pattern recognition

---

**Advantages**

- Works well for linearly separable data
- Reduces dimensionality
- Computationally efficient
- Performs well with small datasets

---

**Limitations**

- Assumes data follows normal distribution
- Assumes equal covariance among classes
- Performs poorly on non-linear data
